```python
import pandas as pd
```

```python
m=pd.read_csv("Mall_Customers.csv")
```

```python
m
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
l=LabelEncoder()
m["Gender"]=l.fit_transform(m["Gender"])
m["Gender"]
```

```python
x=m[["Annual Income (k$)","Spending Score (1-100)"]]
x
```

```python
import scipy.cluster.hierarchy as sch
```

```python
import matplotlib.pyplot as plt
```

```python
s=sch.dendrogram(sch.linkage(x))
plt.show()
```

```python
from sklearn.cluster import AgglomerativeClustering
```

```python
cm=AgglomerativeClustering(n_clusters=5,linkage="ward")
y_pred=cm.fit_predict(x)
```

```python
plt.scatter(x.iloc[:,0],x.iloc[:,1],c=y_pred,cmap="rainbow")
plt.legend()
plt.xlabel("income")
plt.ylabel("spending")
plt.show()
```
