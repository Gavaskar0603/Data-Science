```python
import pandas as pd
```

```python
m=pd.read_csv("Bank-Customers.csv")
```

```python
m
```

```python
x=m[["Earning","Credit Score"]]
x
```

```python
import scipy.cluster.hierarchy as sch
```

```python
import matplotlib.pyplot as plt
```

```python
s=sch.dendrogram(sch.linkage(x))
plt.show()
```

```python
from sklearn.cluster import AgglomerativeClustering
```

```python
cm=AgglomerativeClustering(n_clusters=5,linkage="ward")
y_pre=cm.fit_predict(x)
```

```python
plt.scatter(x.iloc[:,0],x.iloc[:,1],c=y_pre,cmap="rainbow")
plt.legend()
plt.xlabel("Earnings")
plt.ylabel("Credit Score")
plt.show()
```
